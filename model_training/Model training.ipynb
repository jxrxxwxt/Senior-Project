**เปลี่ยนชื่อรูป**

```python
import os

folder_path = "/content/drive/MyDrive/Senior/Main Dataset/Positive Cocci/Staphylococcus saprophyticus 55"
new_name = "Staphylococcus saprophyticus"

#เปลี่ยนชื่อไฟล์
for i, filename in enumerate(sorted(os.listdir(folder_path)), start=1):
    if filename.lower().endswith(".png"):
        old_path = os.path.join(folder_path, filename)
        new_filename = f"{new_name} ({i}).png"
        new_path = os.path.join(folder_path, new_filename)
        os.rename(old_path, new_path)

print("เปลี่ยนชื่อไฟล์เรียบร้อยแล้ว!")
```
